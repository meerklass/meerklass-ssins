```python

import matplotlib.pyplot as plt
import numpy as np

# Define the power law function
def power_law(x, factor, exponent):
    return factor * x ** exponent

def linear_law(x, c, m):
    return m * x + c

def power_fit(amp_rfi, cd_rfi):
    index = np.where((amp_rfi >= 7500) & (amp_rfi <= 100000))
    amp_slope = amp_rfi[index]
    cd_slope = cd_rfi[index]
    x_data = amp_slope
    y_data = cd_slope
    params, covariance = curve_fit(linear_law, np.log(x_data), np.log(y_data))
    fac_fit, exp_fit = np.exp(params[0]), params[1]
    return fac_fit, exp_fit

factor, fit = power_fit(amp_rfi, cd_rfi)
factor_ssins, fit_ssins = power_fit(amp_rfi_ssins, cd_rfi_ssins)

fig, ax = plt.subplots(figsize=(20,12))

ax.plot(amp_rfi, power_law(amp_rfi, factor, fit), color='red', label=f'Fitted Power Law : {fit}')
plt.step(amp_rfi[:-1], cd_rfi, label='RFI detected by the Pipeline', linestyle='--', zorder=5, color='orange')
ax.plot(amp_rfi_ssins, power_law(amp_rfi_ssins, factor_ssins, fit_ssins), color='green', label=f'SSINS Fitted Power Law : {fit_ssins}')
plt.step(amp_rfi_ssins[:-1], cd_rfi_ssins, label='RFI detected by SSINS', linestyle='--', zorder=5, color='k')



ax.set_xlabel('Flux (Jy)')
ax.set_ylabel('Flux Count')
ax.legend()
plt.xscale('log')
plt.yscale('log')
plt.xlim((1e3, 1e6))
plt.title(f'Brightness Distribution of MeerKLASS 2021 L Band ')

# Adjust layout and display the entire grid
plt.tight_layout()
plt.show()
```

```python
def plot_bd(fname, data = None, cd_data = None, amp_data=None, cd=None, amp_rfi=None, cd_rfi=None, amp_norfi=None, cd_norfi=None, label= None, ax=None, figsize= None):
    if ax is None:  # Create a new figure and axes if not being, passed in as a parameter
        fig, ax = plt.subplots(1, 1, figsize=figsize)
 
    if amp_data and cd and amp_rfi and cd_rfi and amp_norfi and cd_norfi is not None:
    
        plt.step(amp_norfi[0:-1], cd_norfi, label='Data not detected as RFI', linestyle='-')
        plt.step(amp_rfi[0:-1], cd_rfi, label='RFI', linestyle='-')
        plt.step(amp_data[0:-1], cd, label='Data', linestyle='-')
        ax.set_xlabel('Flux (Jy)')
        ax.set_ylabel('Flux Count')
        ax.legend()
        plt.xscale('log')
        plt.yscale('log')
        plt.title(f'Brightness Distribution of Observation {fname}')
        plt.xlim((1e3, 1e6))
        plt.show()
    else:
        plt.step(data[0:-1], cd_data, label=label, linestyle='-')
        ax.set_xlabel('Flux (Jy)')
        ax.set_ylabel('Flux Count')
        ax.legend()
        plt.title(f'Brightness Distribution of Observation {fname}')
        plt.xscale('log')
        plt.yscale('log')
        plt.xlim((1e3, 1e6))

         
    return ax
```

```python
for fname, ant_rfi in data.items():
    #print(f'RFI Slope: {fname}')
    obs_cd = []
    obs_amp = []
    #print(obs_cd)
    for ant, value in ant_rfi.items():
        #print(f'{ant}:{value}')
        obs_cd.append(ant_rfi[ant]['cd_rfi'])
        obs_amp.append(ant_rfi[ant]['amp_rfi'])
    cd_rfi = (np.mean(np.array(obs_cd), axis=0))
    amp_rfi = (np.mean(np.array(obs_amp), axis=0))
        
    plot_bd(fname=fname, data=amp_rfi, cd_data=cd_rfi, label='Pipeline RFI')
        
```

```python
import os

# Create output directory for saving the plot
output_dir = "plots"
os.makedirs(output_dir, exist_ok=True)

# Create a figure for all observations
plt.figure(figsize=(24, 12))

for fname, ant_rfi in data_fit.items():
    # Extract antennas and exponents
    ants = list(data_fit[fname].keys())
    exponents = [params['exponent'] for params in data_fit[fname].values()]
    
    # Statistics
    mean_slope = np.mean(exponents)
    std_slope = np.std(exponents)
    
    # Plot for this observation
    plt.scatter(ants, exponents, label=f'{fname} (Mean: {mean_slope:.2f}, STD: {std_slope:.2f})', alpha=0.7)

# Add reference lines (Offringa slopes)
plt.axhline(-1.53, color='r', linestyle='--', linewidth=1, label='Offringa HBA Slope: -1.53')
plt.axhline(-1.62, color='b', linestyle='--', linewidth=1, label='Offringa LBA Slope: -1.62')

# Labels and title
plt.xlabel('Receivers')
plt.ylabel('Slope Exponent')
plt.title('MeerKLASS-SSINS Slope Values vs. Antennas (All Observations)')
plt.xticks(rotation=45, ha='right')
plt.legend()
plt.tight_layout()

# Save the plot
output_path = os.path.join(output_dir, "all_observations_rfi_slope.png")
plt.savefig(output_path, dpi=300, bbox_inches='tight')
print(f"Combined plot saved to: {output_path}")

# Show the plot
plt.show()
```

```python
for fname, ant_rfi in data_fit.items():
    # Create a dictionary of antenna-to-slope mappings
    ant_slope_dict = {ant: params["exponent"] for ant, params in ant_rfi.items()}
    data[f"RFI_Slope_{fname}"] = data["Antenna"].map(ant_slope_dict)
    
    # Determine plot limits with a margin
    east_min, east_max = data["East"].min(), data["East"].max()
    north_min, north_max = data["North"].min(), data["North"].max()
    margin = 50  # Adjust margin as needed

    # Create the plot
    plt.figure(figsize=(20, 12))
    sc = plt.scatter(
        data["East"], data["North"], 
        c=data[f"RFI_Slope_{fname}"], cmap="coolwarm", 
        s=100, edgecolor="k", alpha=0.8
    )
    cb = plt.colorbar(sc, label="RFI Slope Exponent")
    cb.ax.tick_params(labelsize=12)  # Adjust colorbar tick size

    # Add antenna labels
    for _, row in data.iterrows():
        plt.text(row["East"], row["North"], row["Antenna"], 
                 fontsize=8, ha='center', va='center', color='black')
    
    # Set plot limits
    plt.xlim(east_min - margin, east_max + margin)
    plt.ylim(north_min - margin, north_max + margin)
    
    # Customize the plot
    plt.title(f"RFI Slope Exponents Across Antennas for Observation {fname}", fontsize=16)
    plt.xlabel("East (m)", fontsize=14)
    plt.ylabel("North (m)", fontsize=14)
    plt.grid(True, linestyle="--", alpha=0.5)
    
    # Save the plot
    plt.savefig(f"antenna_rfi_slopes_map_{fname}.png", dpi=300, bbox_inches="tight")
    plt.show()
```
