```python

import matplotlib.pyplot as plt
import numpy as np

# Define the power law function
def power_law(x, factor, exponent):
    return factor * x ** exponent

def linear_law(x, c, m):
    return m * x + c

def power_fit(amp_rfi, cd_rfi):
    index = np.where((amp_rfi >= 7500) & (amp_rfi <= 100000))
    amp_slope = amp_rfi[index]
    cd_slope = cd_rfi[index]
    x_data = amp_slope
    y_data = cd_slope
    params, covariance = curve_fit(linear_law, np.log(x_data), np.log(y_data))
    fac_fit, exp_fit = np.exp(params[0]), params[1]
    return fac_fit, exp_fit

factor, fit = power_fit(combined_amp_rfi, combined_cd_rfi)
factor_ssins, fit_ssins = power_fit(combined_amp_rfi_ssins, combined_cd_rfi_ssins)

fig, ax = plt.subplots(figsize=(20,12))

ax.plot(combined_amp_rfi, power_law(combined_amp_rfi, factor, fit), color='red', label=f'Fitted Power Law : {fit}')
plt.step(combined_amp_rfi[:-1], combined_cd_rfi, label='RFI detected by the Pipeline', linestyle='--', zorder=5, color='orange')
ax.plot(combined_amp_rfi_ssins, power_law(combined_amp_rfi_ssins, factor_ssins, fit_ssins), color='green', label=f'SSINS Fitted Power Law : {fit_ssins}')
plt.step(combined_amp_rfi_ssins[:-1], combined_cd_rfi_ssins, label='RFI detected by SSINS', linestyle='--', zorder=5, color='k')


ax.set_xlabel('Amplitude')
ax.set_ylabel('Count')
ax.legend()
plt.xscale('log')
plt.yscale('log')
plt.xlim((1e3, 1.5e6))
plt.title(f'Brightness Distribution from all Recievers in observation {fname}: SSINS Level 0 ')
plt.show()


# Adjust layout and display the entire grid
plt.tight_layout()
plt.show()
```

# Add the fit statistics
1. the slope across the recievers
2. the power law fit for each reciever and for the all antenna distribution

```python

slope_dict = {}

# Define the power law function
def power_law(x, factor, exponent):
    return factor * x ** exponent

def linear_law(x, c, m):
    return m * x + c

# Loop over antennas and fit data

for ant, data in bd_dict.items():
    amp_rfi = data.get('amp_rfi')
    cd_rfi = data.get('cd_rfi')

    # Filter data within the specified range
    index = np.where((amp_rfi >= 8500) & (amp_rfi <= 100000))
    amp_slope = amp_rfi[index]
    cd_slope = cd_rfi[index]

    # Fit the power law
    try:
        x_data = amp_slope
        y_data = cd_slope
        params, covariance = curve_fit(linear_law, np.log(x_data), np.log(y_data))

        # Convert linear parameters to power-law parameters
        fac_fit, exp_fit = np.exp(params[0]), params[1]

        # Save the parameters to the dictionary
        slope_dict[ant] = {
            'factor': fac_fit,
            'exponent': exp_fit,
            'x_data':x_data,
            'y_data':y_data
        }

        # Plot the results for each antenna
        plt.figure(figsize=(10, 6))
        plt.scatter(x_data, y_data, label='Data')
        plt.plot(x_data, power_law(x_data, fac_fit, exp_fit), color='red', label='Fitted Power Law')
        plt.xscale('log')
        plt.yscale('log')
        plt.xlabel('Amplitude (amp_rfi)')
        plt.ylabel('Cross-deviation (cd_rfi)')
        plt.title(f'Power Law Fit for {ant}')
        plt.legend()
        plt.show()

    except Exception as e:
        print(f"Could not fit data for {ant}: {e}")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit
ssins_slope_dict = {}

# Define the power law function
def power_law(x, factor, exponent):
    return factor * x ** exponent

def linear_law(x, c, m):
    return m * x + c

# Loop over antennas and fit data

for ant, data in bd_dict.items():
    amp_rfi = bd_dict[ant]['amp_rfi_ssins']
    cd_rfi = bd_dict[ant]['cd_rfi_ssins']
    amp_rfi = np.array(amp_rfi)
    cd_rfi = np.array(cd_rfi)
    index = np.where((amp_rfi >= 10000) & (amp_rfi <=100000))
    amp_slope = amp_rfi[:-1][index]
    cd_slope = cd_rfi[index]
    # Filter data within the specified range
    

    # Fit the power law
    try:
        x_data = amp_slope
        y_data = cd_slope
        params, covariance = curve_fit(linear_law, np.log(x_data), np.log(y_data))

        # Convert linear parameters to power-law parameters
        fac_fit, exp_fit = np.exp(params[0]), params[1]

        # Save the parameters to the dictionary
        ssins_slope_dict[ant] = {
            'factor': fac_fit,
            'exponent': exp_fit,
            'x_data':x_data,
            'y_data':y_data
        }

        # Plot the results for each antenna
        plt.figure(figsize=(10, 6))
        plt.scatter(x_data, y_data, label='Data')
        plt.plot(x_data, power_law(x_data, fac_fit, exp_fit), color='red', label='Fitted Power Law')
        plt.xscale('log')
        plt.yscale('log')
        plt.xlabel('Amplitude (amp_rfi)')
        plt.ylabel('Cross-deviation (cd_rfi)')
        plt.title(f'Power Law Fit for {ant}')
        plt.legend()
        plt.show()

    except Exception as e:
        print(f"Could not fit data for {ant}: {e}")
```
